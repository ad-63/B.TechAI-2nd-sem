# Regular Expression(Regex):

regular expression is a process of matching.
validating a patteran in a string using `re` module in python 


--> Regex is used in:
1. validating email address
2. validating phone numbers
3. validating URLs
4. validating password size
5. validating seither inputs to the use cases

--> `re` module gives different types of functions for regular expressions.
Functions like:
1. search() ---> search a specific pattern in a given string.
2. match() ---> finds the exact match of a pattern in a test.
3. findall() ---> used to find the provided pattern in a string. 
4. strip() ---> used to seprate text by a specific pattern.
5. sub() ---> replaces the previous value with a new one.

1. search():

```python
import re
text="i love python"
value=re.search("python", text)
print(value)
```

# findall(): 
converts into lists

```python
import re
text="i love python"
val=re.findall("python", text)
print(val)
```

```python
import re
text="i love python"
val=re.findall("python", text)
len_val=len(val)
print(f"python occurs {len_val} times in the given string")
```
