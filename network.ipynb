# Threads:

A thread is a small unit of code that can perform one type of specific task.

--> we can create multiple threads so that it can perform different tasks together at once.
--> in python we use "threading" package to perform thread related program.

eg: A browser can perform multiple tasks togther like:
1. downloading files.
2. writing documents.
3. planing youtube videos.
4. chatting with gf/bf.

# single Threaded program
without using threads

```python
import time
def task1():
    print("task1")
    time.sleep(1)

def task2():
    print("task2")
    time.sleep(1)
task1()
task2()
```

using thread:

```python
import threading
import time 
def task1():
    print("task1")
    time.sleep(1)

t=threading.Thread(target=task1)
t.start()
print("another test performed")
```

Multiple threads


```python
import threading as th
import time
def task1():
    for i in range(5):
        print("task1")
        time.sleep(1)
def task2():
    for i in range(5):
        print("task2")
        time.sleep(1)

t1= th.Thread(target=task1)
t2= th.Thread(target=task2)

t1.start()
t2.start()  
t1.join()
t2.join()
```

# Multithreading:

1. It is a programming approach using which a program can execute multiple tasks concurrently.
2. In multithreading, a program is divided into multiple threads.
3. Each Thread has it's own task and CPU executes one thread at one time.
4. CPU switches between the threads so fast that it seems all threads are executing at the same time.


# Uses:

1. Airline ticketing system where multiple users are accessing the server at the same time.
2. A browser performing multiple tasks within it together.

# Advantages:

1. We can perform many operations together so it saves time.
2. Threads are independent so it doesn't affect other threads if execution occurs in a single thread.

# Disadvantages:

1. Debugging is harder.
2. Unpredictable output order.
