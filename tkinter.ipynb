# compounds and their purpose used in tkinter 

### components_____________________________ purpose
1) label ------------------ display text
2) button ----------------- perform action
3) entry ----------------- single line input
4) text ----------------- multi line input
5) radiobuttons ----------------- single value selection
6) checkbuttons ----------------- multi values selection
7) listboxes ----------------- display a list
8) frame ----------------- creating frames





| Component | Primary Purpose | Selection / Input Style | Key Tkinter Syntax |
| :--- | :--- | :--- | :--- |
| *Label* | Displays static text or images | Read-only / Non-interactive | tk.Label(root, text="...") |
| *Button* | Triggers an action or function when clicked | Single action click | tk.Button(root, command=func) |
| *Entry* | Accepts short input (e.g., username, password) | Single-line text | tk.Entry(root) |
| *Text* | Accepts detailed input (e.g., bio, feedback) | Multi-line text | tk.Text(root, height=5) |
| *Radiobutton* | Selects *1 option only* from a set | Mutually exclusive choice | tk.Radiobutton(root, value=1) |
| *Checkbutton* | Toggles independent options ON or OFF | Multi-value selection | tk.Checkbutton(root, variable=var) |
| *Listbox* | Displays a scrollable list of text items | List selection | tk.Listbox(root) |
| *Frame* | Groups and organizes other widgets | Layout container | tk.Frame(root) |
